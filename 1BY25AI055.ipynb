```python
import itertools
sample_space = list(itertools.product(['H', 'T'], repeat=3))
print("Sample Space:")
print(sample_space)
total = len(sample_space)
exactly_3_heads = [outcome for outcome in sample_space if outcome.count('H') == 3]
exactly_2_heads = [outcome for outcome in sample_space if outcome.count('H') == 2]
at_least_1_head = [outcome for outcome in sample_space if outcome.count('H') >= 1]
p3 = len(exactly_3_heads) / total
p2 = len(exactly_2_heads) / total
p1 = len(at_least_1_head) / total

print("\nProbability of exactly 3 Heads:", p3)
print("Probability of exactly 2 Heads:", p2)
print("Probability of at least 1 Head:", p1)
```

```python
sample_space = [1, 2, 3, 4, 5, 6]

total = len(sample_space)
even = [x for x in sample_space if x % 2 == 0]
odd = [x for x in sample_space if x % 2 != 0]
greater_4 = [x for x in sample_space if x > 4]
less_equal_3 = [x for x in sample_space if x <= 3]

print("Sample Space:", sample_space)

print("Probability of even number:", len(even) / total)
print("Probability of odd number:", len(odd) / total)
print("Probability of number greater than 4:", len(greater_4) / total)
print("Probability of number less than or equal to 3:", len(less_equal_3) / total)
```

```python
import itertools
sample_space = list(itertools.product(range(1, 7), range(1, 7)))

print("Sample Space:")
print(sample_space)

total = len(sample_space)
sum_7 = [outcome for outcome in sample_space
         if outcome[0] + outcome[1] == 7]
sum_greater_8 = [outcome for outcome in sample_space
                 if outcome[0] + outcome[1] > 8]
same_number = [outcome for outcome in sample_space
               if outcome[0] == outcome[1]]
at_least_one_6 = [outcome for outcome in sample_space
                  if outcome[0] == 6 or outcome[1] == 6]
print("\nProbability of sum 7:", len(sum_7) / total)
print("Probability of sum greater than 8:", len(sum_greater_8) / total)
print("Probability of both dice showing same number:", len(same_number) / total)
print("Probability of at least one die showing 6:", len(at_least_one_6) / total)
```

```python
import random
import matplotlib.pyplot as plt

n = 10000
heads = 0
probabilities = []

for i in range(1, n + 1):
    toss = random.choice(['H', 'T'])

    if toss == 'H':
        heads += 1

    probability = heads / i
    probabilities.append(probability)

plt.plot(range(1, n + 1), probabilities)

plt.axhline(y=0.5, linestyle='--', label='P(H) = 0.5')

plt.xlabel("Number of Tosses")
plt.ylabel("Cumulative Probability of Heads")
plt.title("Law of Large Numbers")

plt.xscale('log')
plt.legend()
plt.show()
```

```python
sample_space = [1, 2, 3, 4, 5, 6]

A = [x for x in sample_space if x % 2 == 0]
B = [x for x in sample_space if x > 3]

A_and_B = [x for x in B if x in A]

probability = len(A_and_B) / len(B)

print("A =", A)
print("B =", B)
print("A intersection B =", A_and_B)
print("P(A|B) =", probability)
```

```python
P_Disease = 0.02
P_NotDisease = 1 - P_Disease

P_Positive_given_Disease = 0.95
P_Positive_given_NotDisease = 0.04

P_Positive = (P_Positive_given_Disease * P_Disease) + \
             (P_Positive_given_NotDisease * P_NotDisease)

P_Disease_given_Positive = \
    (P_Positive_given_Disease * P_Disease) / P_Positive

print("P(Disease | Positive) =",
      P_Disease_given_Positive)

print("Percentage =",
      P_Disease_given_Positive * 100, "%")
```

```python
P_Spam = 0.20
P_NotSpam = 0.80

P_Offer_given_Spam = 0.80
P_Offer_given_NotSpam = 0.10

P_Offer = (P_Offer_given_Spam * P_Spam) + \
          (P_Offer_given_NotSpam * P_NotSpam)

P_Spam_given_Offer = \
    (P_Offer_given_Spam * P_Spam) / P_Offer

print("P(Spam | offer) =", P_Spam_given_Offer)
print("Percentage =", P_Spam_given_Offer * 100, "%")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)
n,p = 20, 0.35
binom = stats.binom(n,p)
x = np.arange(0, n+1)
print(f'Mean = {binom.mean():.4f} (Theory: np={n*p})')
```

```python
binom.stats("s")
```

```python
binom.cdf(10)
```

```python
binom.pmf(7)
```

```python
binom.var()
```

```python
binom.moment(2)
```

```python
plt.figure(figsize=(7, 5))
plt.bar(x, binom.pmf(x), color='steelblue',
        edgecolor='black', alpha=0.8)

plt.axvline(binom.mean(), color='red', linestyle='--',
            label=f'Mean = {binom.mean():.1f}')

plt.title(f'Binomial PMF (n={n}, p={p})')
plt.xlabel('Number of Successes (x)')
plt.ylabel('Probability')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
